# Python basics, one cell at a time

A notebook runs code in **cells**. Press **Shift+Enter** (or the ▶ button) to run a cell —
its output appears right under it, and the variables it makes stay available to the cells below.

```python
name = "Warsha"
year = 2026
print("Hello from", name)
```

The last line of a cell is shown automatically — no `print` needed.

```python
year + 1
```

## Lists and loops

Cells can be re-run in any order. Change the list and run it again.

```python
scores = [72, 88, 95, 61]
for s in scores:
    print(s, "passed" if s >= 70 else "try again")
```

```python
average = sum(scores) / len(scores)
round(average, 1)
```

## Functions

A function defined in one cell can be called from any cell after it.

```python
def grade(score):
    if score >= 90:
        return "A"
    if score >= 80:
        return "B"
    if score >= 70:
        return "C"
    return "F"

[grade(s) for s in scores]
```
