# Data with pandas

`scores.csv` sits next to this notebook. pandas reads it into a **DataFrame** — a table you can
filter, sort and summarise. (The first run downloads pandas once; after that it is instant.)

```python
import pandas as pd

df = pd.read_csv("scores.csv")
df.head()
```

## Explore

How big is it, and what does each column look like?

```python
df.shape
```

```python
df.describe()
```

## Filter and sort

Rows where a condition holds, then the best scores first.

```python
passed = df[df["score"] >= 70]
passed.sort_values("score", ascending=False)
```

## Group

Average score per section.

```python
df.groupby("section")["score"].mean().round(1)
```

## Save a result

Files a cell writes appear in the explorer.

```python
passed.to_csv("passed.csv", index=False)
!ls
```
