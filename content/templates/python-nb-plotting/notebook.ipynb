# Plotting with matplotlib

A plot drawn in a cell shows up under it. (matplotlib downloads once, the first time.)

```python
import matplotlib.pyplot as plt

x = list(range(10))
y = [v * v for v in x]

plt.plot(x, y, marker="o")
plt.title("Squares")
plt.xlabel("x")
plt.ylabel("x squared")
plt.show()
```

## Bar chart

Compare a few values side by side.

```python
subjects = ["Math", "Physics", "Programming", "English"]
hours = [6, 4, 8, 3]

plt.bar(subjects, hours, color="#7FC4F5")
plt.title("Study hours this week")
plt.show()
```

## Two lines on one plot

numpy makes whole columns of numbers at once.

```python
import numpy as np

t = np.linspace(0, 2 * np.pi, 200)
plt.plot(t, np.sin(t), label="sin")
plt.plot(t, np.cos(t), label="cos")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
